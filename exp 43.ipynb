```python
import re

# Detect Email and Phone Number
def detect_pii(text):
    email = re.findall(r'[\w\.-]+@[\w\.-]+', text)
    phone = re.findall(r'\b[6-9]\d{9}\b', text)

    return {
        "Email": email,
        "Phone": phone
    }

# Redact PII
def redact(text):
    text = re.sub(r'[\w\.-]+@[\w\.-]+', "[EMAIL_REDACTED]", text)
    text = re.sub(r'\b[6-9]\d{9}\b', "[PHONE_REDACTED]", text)
    return text

# Sample Text
sample = """
Contact: ravi@gmail.com
Phone: 9876543210
"""

print("Original Text:\n", sample)

print("\nDetected PII:")
print(detect_pii(sample))

print("\nRedacted Text:")
print(redact(sample))
```
