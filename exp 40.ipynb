```python
import hashlib

# Calculate SHA-256 hash
def get_hash(data):
    return hashlib.sha256(data.encode()).hexdigest()

# Verify integrity
def verify(data, old_hash):
    return get_hash(data) == old_hash

# Original evidence
evidence = "Invoice payment details"

# Generate hash
hash_value = get_hash(evidence)
print("Original Hash:", hash_value)

# Verify original file
print("Verification:", verify(evidence, hash_value))

# Tampered evidence
tampered = "Invoice payment detail"

# Check tampering
print("After Tampering:", verify(tampered, hash_value))
```
