```python


sender = input("Enter sender email: ")
subject = input("Enter email subject: ")
body = input("Enter email body: ")

suspicious_keywords = [
    "urgent", "verify", "password", "bank", "login",
    "click", "winner", "prize", "free", "account",
    "update", "limited", "confirm"
]

score = 0


if not sender.endswith(("@gmail.com", "@yahoo.com", "@outlook.com", "@company.com")):
    score += 2


for word in suspicious_keywords:
    if word.lower() in subject.lower():
        score += 1


for word in suspicious_keywords:
    if word.lower() in body.lower():
        score += 1


if "http://" in body or "https://" in body:
    score += 2


print("\n--- Email Analysis ---")
print("Suspicion Score:", score)

if score >= 5:
    print("⚠️ Phishing Email Detected!")
else:
    print("✅ Email Appears Safe.")
```
