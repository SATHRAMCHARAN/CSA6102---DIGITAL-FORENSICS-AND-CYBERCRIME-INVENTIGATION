```python
import hashlib
import math
from collections import Counter

def sha256_hash(data: bytes) -> str:
    return hashlib.sha256(data).hexdigest()

def byte_entropy(data: bytes) -> float:
    if not data:
        return 0.0

    counts = Counter(data)
    length = len(data)

    return -sum(
        (c / length) * math.log2(c / length)
        for c in counts.values()
    )

def static_analyze(data: bytes, entropy_threshold=7.5):
    entropy = byte_entropy(data)

    return {
        "sha256": sha256_hash(data),
        "size_bytes": len(data),
        "entropy": round(entropy, 2),
        "likely_packed": entropy >= entropy_threshold,
    }


# Example usage
sample = b"This is a sample file used for malware static analysis."

result = static_analyze(sample)

print(result)
```
