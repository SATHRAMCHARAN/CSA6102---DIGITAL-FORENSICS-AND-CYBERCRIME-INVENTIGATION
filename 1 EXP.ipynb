```python
import os
import time
from datetime import datetime

filename = "evidence.txt"

with open(filename, "w") as file:
    file.write("Original forensic evidence.\n")

def show_metadata(file):
    stats = os.stat(file)
    print("\nFile Metadata")
    print("-" * 30)
    print("Created     :", datetime.fromtimestamp(stats.st_ctime))
    print("Modified    :", datetime.fromtimestamp(stats.st_mtime))
    print("Accessed    :", datetime.fromtimestamp(stats.st_atime))
    print("-" * 30)

print("After File Creation")
show_metadata(filename)

time.sleep(2)

with open(filename, "a") as file:
    file.write("Evidence modified.\n")

print("\nAfter File Modification")
show_metadata(filename)

time.sleep(2)

with open(filename, "r") as file:
    print("\nFile Content:")
    print(file.read())

print("After File Access")
show_metadata(filename)

print("\nLocard's Exchange Principle Demonstrated Successfully.")
```
