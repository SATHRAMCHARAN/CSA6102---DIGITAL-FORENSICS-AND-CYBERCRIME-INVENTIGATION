```python
from datetime import date

# Required fields
required = [
    "case_reference",
    "electronic_record",
    "source_device",
    "hash_value",
    "deponent_name",
    "designation"
]

# Validate fields
def validate(details):
    missing = []
    for field in required:
        if field not in details or details[field] == "":
            missing.append(field)
    return missing

# Generate Certificate
def generate_certificate(details):
    missing = validate(details)

    if missing:
        print("Missing Fields:", missing)
        return

    print("\nCERTIFICATE UNDER SECTION 65B(4)")
    print("-" * 40)
    print("Case Reference :", details["case_reference"])
    print("Date           :", date.today())
    print("Electronic Record :", details["electronic_record"])
    print("Source Device     :", details["source_device"])
    print("SHA-256 Hash      :", details["hash_value"])
    print("Deponent          :", details["deponent_name"])
    print("Designation       :", details["designation"])

# Sample Data
details = {
    "case_reference": "CASE/101",
    "electronic_record": "Laptop Image",
    "source_device": "Dell Laptop",
    "hash_value": "89aba45e26195373d742",
    "deponent_name": "A. Rao",
    "designation": "Scientific Officer"
}

generate_certificate(details)
```
