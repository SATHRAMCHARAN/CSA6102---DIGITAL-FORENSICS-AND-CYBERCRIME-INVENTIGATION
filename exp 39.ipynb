```python
# Simple IT Act 2000 Offence Mapper

IT_ACT = {
    "66D": ["phishing", "fake website", "upi fraud"],
    "66C": ["password", "otp", "identity theft"],
    "65": ["source code", "altered source"],
    "66": ["hacking", "ransomware", "unauthorised access"],
    "66E": ["hidden camera", "private area"],
    "43A": ["body corporate", "sensitive personal data"],
}

def find_section(text):
    text = text.lower()

    for section, keywords in IT_ACT.items():
        for word in keywords:
            if word in text:
                return f"IT Act Section {section} matched (Keyword: {word})"

    return "No matching IT Act section found."

# Example
incident = input("Enter cyber crime description: ")
print(find_section(incident))
```
